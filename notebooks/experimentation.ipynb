# Systemic Risk Exploration and Verification

This notebook visualizes the properties of our Systemic GNN outputs and Optuna Hyperparameters.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import optuna

df = pd.read_csv('../results/predictions.csv')
df.head()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Risk_Score'], bins=50, kde=True)
plt.title('Prediction Distribution of Systemic Risk Scores')
plt.xlabel('Risk Score')
plt.ylabel('Frequency')
plt.show()
```

```python
top_risky = df.head(10)
plt.figure(figsize=(10, 5))
sns.barplot(x='Risk_Score', y='Bank_ID', data=top_risky, palette='Reds_r')
plt.title('Top 10 Riskiest Banks')
plt.show()
```
